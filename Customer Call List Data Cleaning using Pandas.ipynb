### Importing and Viewing the Dataset

### Dropping the Duplicates

```python
df = df.drop_duplicates()
df
#Note the dataset 21 rows changed to 20 since the duplicate has beend dropped(Anaking Skywalker id Duplicate)
```

### Checking the Datatypes of the Dataframe

```python
df.info()
```

### Dropping the Data which isn't useful 

```python
df = df.drop(columns="Not_Useful_Column")
df
```

### Stripping the unnecessary characters in Last Name Column

```python
df["Last_Name"]
```

```python
df["Last_Name"] = df["Last_Name"].str.strip('123._/') 
#Notice Index 2,11,14
df["Last_Name"]
```

### Replacing the Unneccessary Characters

```python
df['Phone_Number']
```

```python
df["Phone_Number"]=df["Phone_Number"].str.replace(r'[^a-zA-Z0-9]', '',regex=True)
df["Phone_Number"]
```

```python
#Simple_Method
df["Phone_Number"] = df["Phone_Number"].astype(str)
#Alternate_method
#df["Phone_Number"].apply(lambda x: str(x))
```

```python
df.info()
# If you see below Phone_number still in object In pandas, strings are stored as object dtype — this is the standard behavior.
# If we apply lambda function it will return without error as shown in next cell
```

```python
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] +'-'+ x[6:10])
df['Phone_Number']
```

### Removing Null Values

```python
df.isnull().sum()
#Note that Last Name and Do not Contact has Null Values
```

```python
df = df.fillna('-')
df
#We can fill anything inside the brackets as per our requirement
#Notice only Last_Name Column has the changes not on Phone_Number since nan is on string type
```

```python
#Remember df["column"].str.replace(pattern, replacement) is the format for .replace()
df["Phone_Number"]=df["Phone_Number"].str.replace('Na','-')
df["Phone_Number"]
```

```python
df["Phone_Number"]=df["Phone_Number"].str.replace('nan','-')
df["Phone_Number"]
```

### Splitting

```python
#Syntax df["column"].str.split(pat, n=-1, expand=False)
#pat	Delimiter — what to split on (,,  , -, etc.)	\s+ (whitespace)
#n	Max splits — how many splits to do (-1 = all)	-1 (all)
#expand	Return DataFrame? — True creates separate columns	False
```

```python
df["Address"].str.split(',',expand=True)
```

```python
df[["Address","State","Zipcode"]]= df["Address"].str.split(',',expand=True)
```

```python
df
```

### Formatting the Yes or NO to Y or N 

#### Multiple columns, single replacement
#### df[['col1', 'col2']].replace('old', 'new')

#### Multiple columns, multiple replacements
#### df[['col1', 'col2']].replace({'old1': 'new1', 'old2': 'new2'})

```python
df[["Paying Customer","Do_Not_Contact"]] = df[["Paying Customer","Do_Not_Contact"]].replace('Yes','Y')
df[["Paying Customer","Do_Not_Contact"]]
#Notice that we didnt use str.replace() but.replace() instead because we are working on a Dataframe and not on a string column. 
# If we use str.replace() it will give us an error since it is not a string column but a dataframe.
```

```python
df[["Paying Customer","Do_Not_Contact"]] = df[["Paying Customer","Do_Not_Contact"]].replace('No','N')
df[["Paying Customer","Do_Not_Contact"]]
```

```python
df
```

```python
df.fillna('-')
```

```python
# Here we do not want to call customer with Do_Not_Contact as Y and Phone_Number  as - since it is not a valid phone number. 
#But if we use dropna() it will drop the entire row which is not what we want.
```

```python
for x in df.index:
    if df.loc[x, 'Do_Not_Contact'] == 'Y':
        df.drop(x, inplace=True)
```

```python
for x in df.index:
    if df.loc[x, 'Phone_Number'] == '---':
        df.drop(x, inplace=True)
df
```
