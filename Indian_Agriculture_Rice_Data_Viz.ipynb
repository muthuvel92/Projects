### RICE STATE

```python
rice_state = df.groupby('State Name')['RICE PRODUCTION (1000 tons)'].sum().sort_values(ascending=False).reset_index().head(10)
rice_state
```

```python
plt.figure(figsize=(25, 5))
plt.title('Top 10 rice productions by States')
sns.barplot(data = rice_state, x='State Name', y='RICE PRODUCTION (1000 tons)',palette='magma')
```

### Using Plotly for Interactive data

```python
px.bar(rice_state, x='State Name', y ='RICE PRODUCTION (1000 tons)', hover_data='RICE PRODUCTION (1000 tons)',color ='RICE PRODUCTION (1000 tons)',color_continuous_scale='viridis',title='Top 10 Rice Production States')
```

```python
rice_area = df.groupby('State Name')['RICE AREA (1000 ha)'].sum().sort_values(ascending=False).reset_index().head(10)
rice_area
```

```python
px.scatter(rice_area, 
           x= 'State Name', 
           y ='RICE AREA (1000 ha)',
           color = 'RICE AREA (1000 ha)',
           size ='RICE AREA (1000 ha)', 
           hover_data='RICE AREA (1000 ha)', 
           title= 'Top 10 Rice Area',)
```

```python
ten_year_riceprod = df.groupby('Year')['RICE PRODUCTION (1000 tons)'].sum().sort_values(ascending=False).reset_index().head(10)
ten_year_riceprod
```

```python
fig_rice = px.bar(
    ten_year_riceprod,
    x='Year',
    y='RICE PRODUCTION (1000 tons)',
    color ='RICE PRODUCTION (1000 tons)',
    color_continuous_scale='viridis',
    title=' Top 10 Year Rice Production',
template='plotly_dark'
)
fig_rice
```

```python
#Since Year is in int64 Plotly considers as Continuous axis and arranges numerically, If we can convert into string for categorical purposes it will sort the issue
top10 = ten_year_riceprod.copy()

top10['Year'] = top10['Year'].astype(str)

fig = px.bar(
    top10,
    x='Year',
    y='RICE PRODUCTION (1000 tons)',
    color='RICE PRODUCTION (1000 tons)',
    color_continuous_scale='viridis',
    title='Top 10 Year Rice Production',
    template='plotly_dark'
)

fig.show()
```

```python
rice_yield = df.groupby('State Name')['RICE YIELD (Kg per ha)'].sum().sort_values(ascending=False).reset_index().head(10)
rice_yield
```

```python
top_10_yield = px.pie(
    rice_yield,
    values='RICE YIELD (Kg per ha)',
    names='State Name',
    title='Top 10 Rice Yield percentage',
    template='plotly_dark'
    )
top_10_yield
```
